```python
import shutil
from pathlib import Path

src_dir = Path("/content/drive/MyDrive/building_LLM/data")
dst_dir = Path("/content")

for name in (
    "TinyStoriesV2-GPT4-train.txt",
    "TinyStoriesV2-GPT4-valid.txt",
):
    src = src_dir / name
    dst = dst_dir / name
    if not src.is_file():
        raise FileNotFoundError(f"Not found: {src}")
    print(f"Copying {name} …")
    shutil.copy2(src, dst)
    print(f"  -> {dst} ({dst.stat().st_size / 1e9:.2f} GB)")
```

```python
import shutil
from pathlib import Path
from datetime import datetime

drive_root = Path("/content/drive/MyDrive/building_LLM")
run_dir = drive_root / f"run_{datetime.now().strftime('%Y%m%d_%H%M%S')}"
run_dir.mkdir(parents=True, exist_ok=True)

# Checkpoints (best.pt, latest.pt, etc.)
shutil.copytree("/content/checkpoints", run_dir / "checkpoints", dirs_exist_ok=True)

# Optional: token caches so future runs skip long encode
for name in ("cache_train.pt", "cache_val.pt"):
    src = Path("/content") / name
    if src.is_file():
        shutil.copy2(src, run_dir / name)
        print("saved", name)
    else:
        print("skip (missing)", src)

# Optional: copy tokenizer you used
tok_src = Path("/content/custom_llm/tokenizer")
if (tok_src / "tokenizer.json").is_file():
    shutil.copytree(tok_src, run_dir / "tokenizer", dirs_exist_ok=True)

print("Saved to:", run_dir)
```
